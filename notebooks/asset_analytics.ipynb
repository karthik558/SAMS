# SAMS Asset Analytics and Operational Modeling

This notebook provides analytical modeling and visual graphs for the Smart Asset Management System (SAMS):
1. Asset category distribution and valuation
2. Straight-Line Depreciation (SLD) curves over asset useful lifespan
3. Maintenance ticket flow and resolution velocity
4. Annual Maintenance Contract (AMC) expiration risk timelines

```python
import matplotlib.pyplot as plt
import numpy as np

# Set clean visualization style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = 'Helvetica, Arial, sans-serif'
plt.rcParams['axes.edgecolor'] = '#e2e8f0'
plt.rcParams['axes.linewidth'] = 0.8
```

## 1. Asset Distribution and Portfolio Valuation

Visualizing asset volume and average acquisition cost across core enterprise categories.

```python
categories = ['IT Equipment', 'Office Furniture', 'Machinery & HVAC', 'Vehicles', 'Network & Telecom']
counts = [142, 85, 38, 14, 46]
avg_cost_inr = [65000, 18000, 120000, 450000, 180000]

fig, ax1 = plt.subplots(figsize=(10, 5))

x = np.arange(len(categories))
width = 0.38

rects1 = ax1.bar(x - width/2, counts, width, label='Asset Count', color='#2563eb', alpha=0.9)
ax1.set_ylabel('Total Units', color='#2563eb', fontsize=11, fontweight='bold')
ax1.tick_params(axis='y', labelcolor='#2563eb')
ax1.set_xticks(x)
ax1.set_xticklabels(categories, rotation=15, ha='right', fontsize=10)

ax2 = ax1.twinx()
rects2 = ax2.bar(x + width/2, [c / 1000 for c in avg_cost_inr], width, label='Avg Cost (k INR)', color='#10b981', alpha=0.85)
ax2.set_ylabel('Avg Unit Cost (Thousands INR)', color='#10b981', fontsize=11, fontweight='bold')
ax2.tick_params(axis='y', labelcolor='#10b981')
ax2.grid(False)

plt.title('Asset Portfolio: Volume vs. Average Unit Acquisition Value', fontsize=13, fontweight='bold', pad=15)
fig.tight_layout()
plt.show()
```

## 2. Straight-Line Depreciation (SLD) Model

Simulating asset value amortization over a 5-year useful life cycle:
- Acquisition Cost: INR 100,000
- Salvage Value (5%): INR 5,000
- Depreciable Base: INR 95,000
- Annual Depreciation: INR 19,000 / year

```python
years = np.linspace(0, 5, 61)
cost = 100000
salvage = 5000
depreciable = cost - salvage
annual_rate = depreciable / 5.0

accum_deprec = np.minimum(depreciable, annual_rate * years)
book_value = cost - accum_deprec

plt.figure(figsize=(9, 5))
plt.plot(years, book_value, label='Net Book Value (Remaining Asset Worth)', color='#059669', linewidth=2.5)
plt.plot(years, accum_deprec, label='Accumulated Depreciation', color='#dc2626', linewidth=2.2, linestyle='--')
plt.axhline(y=salvage, color='#64748b', linestyle=':', label=f'Salvage Floor (INR {salvage:,})')

plt.title('Straight-Line Depreciation Lifecycle (5-Year Class)', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Elapsed Operational Time (Years)', fontsize=11)
plt.ylabel('Asset Value (INR)', fontsize=11)
plt.legend(frameon=True, facecolor='#f8fafc', edgecolor='#cbd5e1')
plt.tight_layout()
plt.show()
```

## 3. Maintenance Ticket Flow & Resolution Velocity

Tracking monthly incoming service requests vs. completed resolutions to monitor backlog health.

```python
months = ['Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov']
created = [24, 32, 28, 41, 35, 29]
resolved = [20, 29, 30, 38, 36, 31]
backlog = [4, 7, 5, 8, 7, 5]

plt.figure(figsize=(9, 4.8))
plt.fill_between(months, created, alpha=0.18, color='#0284c7')
plt.plot(months, created, marker='o', label='Tickets Created', color='#0284c7', linewidth=2)
plt.fill_between(months, resolved, alpha=0.18, color='#10b981')
plt.plot(months, resolved, marker='s', label='Tickets Resolved', color='#10b981', linewidth=2)
plt.plot(months, backlog, marker='^', label='Pending Backlog', color='#f59e0b', linewidth=2, linestyle=':')

plt.title('Monthly Maintenance Velocity (Created vs. Resolved)', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Month', fontsize=11)
plt.ylabel('Ticket Count', fontsize=11)
plt.legend(loc='upper left', frameon=True, facecolor='#f8fafc', edgecolor='#cbd5e1')
plt.tight_layout()
plt.show()
```
